```python
import pandas as pd

flows = pd.read_csv("../data/my_flows.csv")
print("Flows found:", len(flows))
print("Columns:", len(flows.columns))
flows.head()
```

```python
list(flows.columns)
```

```python
import json, re

with open("../models/feature_columns.json") as f:
    model_features = json.load(f)

def normalize(name):
    return re.sub(r"[^a-z0-9]", "", name.lower())

tool_lookup = {normalize(c): c for c in flows.columns}

matched, missing = {}, []
for feat in model_features:
    key = normalize(feat)
    if key in tool_lookup:
        matched[feat] = tool_lookup[key]
    else:
        missing.append(feat)

print("Model needs:", len(model_features), "features")
print("Matched automatically:", len(matched))
print("Not matched:", len(missing))
```

```python
print("MODEL features with no match:")
for m in missing:
    print("  ", m)

print()
print("TOOL columns not used yet:")
used = set(matched.values())
for c in flows.columns:
    if c not in used:
        print("  ", c)
```

```python
import difflib

tool_unused = [c for c in flows.columns if c not in set(matched.values())]
tool_unused_norm = {normalize(c): c for c in tool_unused}

for m in missing:
    guesses = difflib.get_close_matches(normalize(m), tool_unused_norm.keys(), n=2, cutoff=0.5)
    options = [tool_unused_norm[g] for g in guesses]
    print(f"{m:45s} -> {options}")
```

```python
manual_map = {
    # "model feature name": "tool column name",
}
matched.update({k: v for k, v in manual_map.items() if v in flows.columns})
print("Total matched now:", len(matched))
print("Still missing:", len(model_features) - len(matched))
```

```python
train_df = pd.read_csv("../data/Wednesday-workingHours.pcap_ISCX.csv")
train_df.columns = train_df.columns.str.strip()

rows = []
for model_name, tool_name in matched.items():
    if model_name in train_df.columns:
        rows.append({
            "feature": model_name,
            "dataset_median": train_df[model_name].median(),
            "your_pcap_median": flows[tool_name].median(),
        })

compare = pd.DataFrame(rows)
compare
```

```python
import json
with open("../models/feature_mapping.json", "w") as f:
    json.dump(matched, f, indent=2)
print("Saved", len(matched), "matches")
```
